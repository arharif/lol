# Question 1 — LU Factorization

An LU factorization writes a matrix as

$$
\boxed{A=LU}
$$

where:

- $$L$$ is lower triangular;
- $$U$$ is upper triangular.

The elimination multipliers used during Gaussian elimination are stored in $$L$$.

## 1(a)

Given

$$
A=
\begin{pmatrix}
2&5\\
-3&-4
\end{pmatrix}.
$$

Eliminate the entry below the first pivot.

The multiplier is

$$
m_{21}
=
\frac{-3}{2}.
$$

So

$$
L=
\begin{pmatrix}
1&0\\
-\frac32&1
\end{pmatrix}.
$$

Applying

$$
R_2\leftarrow R_2-\left(-\frac32\right)R_1
$$

gives

$$
U=
\begin{pmatrix}
2&5\\
0&\frac72
\end{pmatrix}.
$$

Therefore,

$$
\boxed{
A=
\begin{pmatrix}
1&0\\
-\frac32&1
\end{pmatrix}
\begin{pmatrix}
2&5\\
0&\frac72
\end{pmatrix}
}
$$

## 1(b)

Given

$$
A=
\begin{pmatrix}
3&-7&-2\\
-3&5&1\\
6&-4&0
\end{pmatrix}.
$$

The first elimination multipliers are

$$
m_{21}=-1,
\qquad
m_{31}=2.
$$

After the first column is eliminated, the second pivot is $$-2$$.

The next multiplier is

$$
m_{32}=-5.
$$

Thus,

$$
L=
\begin{pmatrix}
1&0&0\\
-1&1&0\\
2&-5&1
\end{pmatrix},
$$

and

$$
U=
\begin{pmatrix}
3&-7&-2\\
0&-2&-1\\
0&0&-1
\end{pmatrix}.
$$

Therefore,

$$
\boxed{A=LU}
$$

with the matrices above.

````python
A1a = sp.Matrix([[2, 5], [-3, -4]])
A1b = sp.Matrix([[3, -7, -2], [-3, 5, 1], [6, -4, 0]])

for name, A in [("1(a)", A1a), ("1(b)", A1b)]:
    L, U, _ = A.LUdecomposition()
    print(name)
    print("L =")
    sp.pprint(L)
    print("U =")
    sp.pprint(U)
    print("Check L*U = A:", L*U == A)
    print()
````

# Question 2 — Least-Squares Solution

Given

$$
A=
\begin{pmatrix}
1&-2\\
-1&2\\
0&3\\
2&5
\end{pmatrix},
\qquad
b=
\begin{pmatrix}
3\\
1\\
-4\\
2
\end{pmatrix}.
$$

Because the system is overdetermined, an exact solution may not exist.

The least-squares solution minimizes

$$
\|Ax-b\|.
$$

It satisfies the **normal equations**

$$
\boxed{
A^TA\hat{x}=A^Tb.
}
$$

Solving gives

$$
\boxed{
\hat{x}
=
\begin{pmatrix}
\frac43\\
-\frac13
\end{pmatrix}.
}
$$

The residual is

$$
r=A\hat{x}-b
=
\begin{pmatrix}
-1\\
-3\\
3\\
-1
\end{pmatrix}.
$$

The least-squares error is usually the residual norm:

$$
\|r\|
=
\sqrt{(-1)^2+(-3)^2+3^2+(-1)^2}
=
\sqrt{20}
=
2\sqrt5.
$$

Therefore,

$$
\boxed{
\text{least-squares error}=2\sqrt5
}
$$

and the squared error is

$$
\boxed{20}.
$$

````python
A2 = sp.Matrix([
    [1, -2],
    [-1, 2],
    [0, 3],
    [2, 5]
])

b2 = sp.Matrix([3, 1, -4, 2])

xhat = (A2.T * A2).inv() * A2.T * b2
residual = A2 * xhat - b2

print("Least-squares solution:")
sp.pprint(xhat)

print("\nResidual:")
sp.pprint(residual)

print("\nResidual norm:")
sp.pprint(sp.sqrt((residual.T * residual)[0]))
````

# Question 3 — Best-Fit Line

Data points:

$$
(2,3),\quad(3,2),\quad(5,1),\quad(6,0).
$$

We model the data by

$$
y=mx+b.
$$

The design matrix is

$$
X=
\begin{pmatrix}
2&1\\
3&1\\
5&1\\
6&1
\end{pmatrix},
$$

the parameter vector is

$$
\beta=
\begin{pmatrix}
m\\b
\end{pmatrix},
$$

and

$$
y=
\begin{pmatrix}
3\\2\\1\\0
\end{pmatrix}.
$$

The least-squares solution is

$$
\hat{\beta}
=
(X^TX)^{-1}X^Ty.
$$

This gives

$$
m=-\frac7{10},
\qquad
b=\frac{43}{10}.
$$

Therefore,

$$
\boxed{
y=-0.7x+4.3
}
$$

````python
points3 = np.array([[2, 3], [3, 2], [5, 1], [6, 0]], dtype=float)

X3 = np.column_stack((points3[:, 0], np.ones(len(points3))))
y3 = points3[:, 1]

beta3 = np.linalg.lstsq(X3, y3, rcond=None)[0]
m3, b3 = beta3

print("m =", m3)
print("b =", b3)

xx = np.linspace(1.5, 6.5, 200)

plt.figure(figsize=(6, 4))
plt.scatter(points3[:, 0], points3[:, 1], label="Data")
plt.plot(xx, m3 * xx + b3, label="Least-squares line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Question 3: Least-Squares Best-Fit Line")
plt.grid(True)
plt.legend()
plt.show()
````

# Question 5 — Polynomial Least Squares

The model is

$$
y=b_1x+b_2x^2+b_3x^3.
$$

Notice there is **no constant term**.

## 5(a) Design matrix and parameter vector

For data

$$
(x_1,y_1),\ldots,(x_n,y_n),
$$

the model gives

$$
y_i=b_1x_i+b_2x_i^2+b_3x_i^3.
$$

Therefore the design matrix is

$$
\boxed{
X=
\begin{pmatrix}
x_1&x_1^2&x_1^3\\
x_2&x_2^2&x_2^3\\
\vdots&\vdots&\vdots\\
x_n&x_n^2&x_n^3
\end{pmatrix}
}
$$

and the parameter vector is

$$
\boxed{
\beta=
\begin{pmatrix}
b_1\\
b_2\\
b_3
\end{pmatrix}.
}
$$

The least-squares problem is

$$
X\beta\approx y.
$$

## 5(b) Fit the data

The data are

$$
(4,1.58),(6,2.08),(8,2.5),(10,2.8),
$$

$$
(12,3.2),(14,3.3),(16,3.9),(18,4.31).
$$

Using NumPy least squares gives approximately

$$
b_1=0.50933042,
$$

$$
b_2=-0.03259626,
$$

$$
b_3=0.00098071.
$$

Therefore the fitted curve is

$$
\boxed{
y
\approx
0.50933042x
-0.03259626x^2
+0.00098071x^3.
}
$$

````python
x5 = np.array([4, 6, 8, 10, 12, 14, 16, 18], dtype=float)
y5 = np.array([1.58, 2.08, 2.5, 2.8, 3.2, 3.3, 3.9, 4.31], dtype=float)

X5 = np.column_stack((x5, x5**2, x5**3))

b1, b2, b3 = np.linalg.lstsq(X5, y5, rcond=None)[0]

print("b1 =", b1)
print("b2 =", b2)
print("b3 =", b3)

print(f"y = {b1:.8f}x + ({b2:.8f})x^2 + ({b3:.8f})x^3")
````

## 5(c) Plot the data and least-squares curve

````python
xx5 = np.linspace(x5.min(), x5.max(), 400)
yy5 = b1*xx5 + b2*xx5**2 + b3*xx5**3

plt.figure(figsize=(7, 5))
plt.scatter(x5, y5, label="Data points")
plt.plot(xx5, yy5, label="Least-squares cubic")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Question 5: Polynomial Least-Squares Fit")
plt.grid(True)
plt.legend()
plt.show()
````

# Question 7 — SVD of a $$2\times2$$ Matrix

Given

$$
A=
\begin{pmatrix}
2&-1\\
2&2
\end{pmatrix}.
$$

We want

$$
\boxed{A=U\Sigma V^T.}
$$

The eigenvalues of $$A^TA$$ are

$$
9,\qquad4,
$$

so the singular values are

$$
3,\qquad2.
$$

A valid SVD is

$$
U=
\frac1{\sqrt5}
\begin{pmatrix}
1&-2\\
2&1
\end{pmatrix},
$$

$$
\Sigma=
\begin{pmatrix}
3&0\\
0&2
\end{pmatrix},
$$

and

$$
V=
\frac1{\sqrt5}
\begin{pmatrix}
2&-1\\
1&2
\end{pmatrix}.
$$

Thus,

$$
\boxed{
A=U\Sigma V^T.
}
$$

Different sign choices for matching singular vectors can produce another equally valid SVD.

````python
A7 = np.array([[2, -1], [2, 2]], dtype=float)

U7, s7, VT7 = np.linalg.svd(A7)

print("U =")
print(U7)
print("\nSingular values =", s7)
print("\nV^T =")
print(VT7)
print("\nReconstruction =")
print(U7 @ np.diag(s7) @ VT7)
````
