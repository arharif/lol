
# Question 2.1 - `CreateList(n)`

### What is required?

Create a function that returns every integer from \(1\) through \(3n+1\) that has remainder \(1\) when divided by \(3\).

These numbers follow the pattern:

$$
1,\;4,\;7,\;10,\;\ldots,\;3n+1
$$

The difference between consecutive values is always \(3\).

Python's `range(start, stop, step)` is therefore a natural solution.

Because the `stop` value of `range` is **excluded**, we use:

```python
range(1, 3*n + 2, 3)
```

so that \(3n+1\) is included.


````python

def CreateList(n):
    return list(range(1, 3 * n + 2, 3))
````

````python

# Test
print(CreateList(4))
````


### Observation

For \(n=4\),

\[
3n+1=13
\]

and the result is:

```text
[1, 4, 7, 10, 13]
```

Every number leaves remainder \(1\) after division by \(3\).



# Question 2.2 — `ShowConcatenation(n, m)`

The function must return a list containing **three lists**.

### First list — even numbers

Start at $4$ and end at $2n+4$:

$$
4,\;6,\;8,\;\ldots,\;2n+4
$$

The step is $2$.

### Second list — odd numbers

Start at $501$ and end at $2m+501$:

$$
501,\;503,\;505,\;\ldots,\;2m+501
$$

Again, the step is \(2\).

### Third list — concatenation

Concatenation means placing the second list immediately after the first:

```python
even_list + odd_list
```

The required return structure is therefore:

```python
[even_list, odd_list, even_list + odd_list]
```


````python

def ShowConcatenation(n, m):
    even_list = list(range(4, 2 * n + 5, 2))
    odd_list = list(range(501, 2 * m + 502, 2))
    return [even_list, odd_list, even_list + odd_list]
````

````python

# Test
print(ShowConcatenation(3, 2))
````


### Observation

For `n = 3` and `m = 2`:

- even list: `[4, 6, 8, 10]`
- odd list: `[501, 503, 505]`
- concatenation: `[4, 6, 8, 10, 501, 503, 505]`

Notice that concatenation does **not** add the numbers mathematically.  
It joins the two lists.



# Question 2.3 — `Sigmoid(x)`

The logistic sigmoid function is:

$$
S(x) = \frac{1}{1+e^{-x}}
$$

It converts a real number into a value between $0$ and $1$.

For example:

$$
S(0) = \frac{1}{2} = 0.5
$$

The assignment asks us to **return** the result, not merely print it.

The implementation below uses an equivalent numerically stable form for negative inputs. This avoids overflow for very large negative values while computing the same sigmoid function.


````python

def Sigmoid(x):
    if x >= 0:
        return 1 / (1 + math.exp(-x))

    exp_x = math.exp(x)
    return exp_x / (1 + exp_x)
````

````python

# Tests
print("Sigmoid(0) =", Sigmoid(0))
print("Sigmoid(2) =", Sigmoid(2))
print("Sigmoid(-2) =", Sigmoid(-2))
````


### Observation

- Negative inputs produce values closer to \(0\).
- \(x=0\) produces exactly \(0.5\).
- Positive inputs produce values closer to \(1\).

This is why sigmoid is useful as a probability-like output transformation in binary classification.



# Question 2.4 — `Av(L)`

The function must calculate the arithmetic mean of a list:

\[
\text{mean}
=
\frac{x_1+x_2+\cdots+x_n}{n}
\]

For example:

```python
[2, 4, 6]
```

has mean:

\[
\frac{2+4+6}{3}=4
\]

However, the assignment adds two important rules.

The function must return `None` if:

1. the list is empty;
2. at least one element is not numeric.

For this assignment, we treat Python `int` and `float` values as numeric. Boolean values are rejected because they are logical values rather than intended numerical observations.


````python

def Av(L):
    if len(L) == 0:
        return None

    for value in L:
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            return None

    return sum(L) / len(L)
````

````python

# Tests
print(Av([2, 4, 6]))
print(Av([1.5, 2.5, 3.5]))
print(Av([]))
print(Av([1, "hello", 3]))
````


### Observation

Expected behavior:

```text

2.5
None
None
```

The validation must happen **before** calculating the average.

# Question 2.6 — `GD(f, x0, n, eta)`

This is the most important function mathematically.

`GD` implements **gradient descent in one variable**.

The update rule is:

$$
x_{i+1} = x_i - \eta f'(x_i)
$$

where:

- \(f\) is the function we want to minimise;
- \(f'(x)\) is its derivative;
- \(x_0\) is the initial guess;
- \(n\) is the number of sequence terms to return;
- \(\eta\) (`eta`) is the learning rate.

### Why subtract the derivative?

The derivative points in the direction in which the function increases.

Therefore:

\[
-f'(x)
\]

points toward decrease.

Gradient descent repeatedly takes a step in that downhill direction.

### Role of SymPy

If the input is:

```python
f = x**2
```

then SymPy computes:

\[
f'(x)=2x
\]

using:

```python
fp = f.diff(x)
```

`lambdify` then converts the symbolic expressions into ordinary callable Python functions.


````python

def GD(f, x0, n, eta):
    fp = f.diff(x)
    f = lambdify(x, f)
    fp = lambdify(x, fp)

    L = [x0]

    for _ in range(n - 1):
        x_next = L[-1] - eta * fp(L[-1])
        L.append(x_next)

    return L
````

The update becomes:

\[
x_{i+1}=x_i-0.1(2x_i)=0.8x_i.
\]

So every new term is \(80\%\) of the preceding term.

````python
import random
import math
from sympy import Symbol, lambdify

x = Symbol("x")


def CreateList(n):
    return list(range(1, 3 * n + 2, 3))


def ShowConcatenation(n, m):
    even_list = list(range(4, 2 * n + 5, 2))
    odd_list = list(range(501, 2 * m + 502, 2))
    return [even_list, odd_list, even_list + odd_list]


def Sigmoid(x):
    if x >= 0:
        return 1 / (1 + math.exp(-x))
    exp_x = math.exp(x)
    return exp_x / (1 + exp_x)


def Av(L):
    if len(L) == 0:
        return None

    for value in L:
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            return None

    return sum(L) / len(L)


def RandAv(n):
    L = [random.randint(1, n) for _ in range(n)]
    # print(L)  # Uncomment only if your instructor wants the Section 2.5 print behavior.
    return Av(L)


def GD(f, x0, n, eta):
    fp = f.diff(x)
    f = lambdify(x, f)
    fp = lambdify(x, fp)

    L = [x0]

    for _ in range(n - 1):
        x_next = L[-1] - eta * fp(L[-1])
        L.append(x_next)

    return L
````
