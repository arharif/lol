# Question 1 — Independent Binomial Random Variables

The assignment gives

$$
X_1\sim b\left(3,\frac12\right)
$$

and

$$
X_2\sim b\left(5,\frac12\right),
$$

and states that $X_1$ and $X_2$ are independent.

For a binomial random variable,

$$
P(X=k)
=
\binom{n}{k}p^k(1-p)^{n-k}.
$$

Because $p=1/2$ here,

$$
P(X=k)
=
\binom{n}{k}\left(\frac12\right)^n.
$$

## 1(a) Find $P(X_1=2,\;X_2=4)$

Because $X_1$ and $X_2$ are independent,

$$
P(X_1=2,\;X_2=4)
=
P(X_1=2)P(X_2=4).
$$

First,

$$
P(X_1=2)
=
\binom32
\left(\frac12\right)^3
=
3\cdot\frac18
=
\frac38.
$$

Next,

$$
P(X_2=4)
=
\binom54
\left(\frac12\right)^5
=
5\cdot\frac1{32}
=
\frac5{32}.
$$

Therefore,

$$
P(X_1=2,\;X_2=4)
=
\frac38\cdot\frac5{32}
=
\frac{15}{256}.
$$

So,

$$
\boxed{
P(X_1=2,\;X_2=4)
=
\frac{15}{256}
\approx0.05859375
}
$$

## 1(b) Find $P(X_1+X_2=7)$

Since the variables are independent and have the same success probability $p=1/2$,

$$
X_1+X_2
\sim
b\left(3+5,\frac12\right)
=
b\left(8,\frac12\right).
$$

Therefore,

$$
P(X_1+X_2=7)
=
\binom87
\left(\frac12\right)^8.
$$

Thus,

$$
=
8\cdot\frac1{256}
=
\frac8{256}
=
\frac1{32}.
$$

Hence,

$$
\boxed{
P(X_1+X_2=7)
=
\frac1{32}
=
0.03125
}
$$

### Direct check

$$

# Question 4 — Mean and Variance of a Sum

The assignment gives a random sample of size 2 from the pdf

$$
f(x)=6x(1-x),
\qquad
0<x<1.
$$

Let

$$
Y=X_1+X_2.
$$

Because $X_1$ and $X_2$ form a random sample, they are independent and identically distributed.

We first find the mean and variance of one observation $X$.

## Step 1 — Find $E[X]$

$$
E[X]
=
\int_0^1x\,6x(1-x)\,dx.
$$

Expand:

$$
E[X]
=
6\int_0^1(x^2-x^3)\,dx.
$$

Therefore,

$$
=
6\left(
\frac13-\frac14
\right)
=
6\left(\frac1{12}\right)
=
\frac12.
$$

Thus,

$$
E[X]=\frac12.
$$

## Step 2 — Find $E[X^2]$

$$
E[X^2]
=
\int_0^1x^2\,6x(1-x)\,dx.
$$

Therefore,

$$
=
6\int_0^1(x^3-x^4)\,dx.
$$

So,

$$
=
6\left(
\frac14-\frac15
\right)
=
6\left(\frac1{20}\right)
=
\frac3{10}.
$$

## Step 3 — Find $\operatorname{Var}(X)$

$$
\operatorname{Var}(X)
=
E[X^2]-E[X]^2.
$$

Thus,

$$
=
\frac3{10}
-
\left(\frac12\right)^2
=
\frac3{10}-\frac14.
$$

Using denominator 20,

$$
=
\frac6{20}-\frac5{20}
=
\frac1{20}.
$$

## Step 4 — Find the mean of $Y$

$$
E[Y]
=
E[X_1+X_2].
$$

By linearity,

$$
E[Y]
=
E[X_1]+E[X_2]
=
\frac12+\frac12
=
1.
$$

Therefore,

$$
\boxed{E[Y]=1}.
$$

## Step 5 — Find the variance of $Y$

Since $X_1$ and $X_2$ are independent,

$$
\operatorname{Var}(Y)
=
\operatorname{Var}(X_1)+\operatorname{Var}(X_2).
$$

Thus,

$$
=
\frac1{20}+\frac1{20}
=
\frac1{10}.
$$

Therefore,

$$
\boxed{
\operatorname{Var}(Y)=\frac1{10}=0.1
}
$$

# Question 6 — Central Limit Theorem for a Sample Mean

The professor explicitly says to use the **Central Limit Theorem** for Questions 6–8.

The population pdf is

$$
f(x)
=
\frac32x^2,
\qquad
-1<x<1.
$$

A random sample of size

$$
n=15
$$

is taken, and $\bar X$ denotes the sample mean.

We want

$$
P(-0.03<\bar X<0.15).
$$

To use the CLT, first find the population mean and variance.

## Step 1 — Population mean

$$
\mu
=
E[X]
=
\int_{-1}^{1}
x\left(\frac32x^2\right)dx.
$$

Thus,

$$
\mu
=
\frac32
\int_{-1}^{1}x^3\,dx.
$$

Because $x^3$ is an odd function integrated symmetrically over $[-1,1]$,

$$
\boxed{\mu=0}.
$$

## Step 2 — Population variance

Since $\mu=0$,

$$
\operatorname{Var}(X)=E[X^2].
$$

Now,

$$
E[X^2]
=
\int_{-1}^{1}
x^2\left(\frac32x^2\right)dx.
$$

Therefore,

$$
=
\frac32
\int_{-1}^{1}x^4\,dx.
$$

Since $x^4$ is even,

$$
=
\frac32
\left(
2\int_0^1x^4\,dx
\right).
$$

Thus,

$$
=
\frac32
\left(
2\cdot\frac15
\right)
=
\frac35.
$$

Therefore,

$$
\boxed{
\sigma^2=\frac35
}.
$$

## Step 3 — Sampling distribution of $\bar X$

By the Central Limit Theorem,

$$
\bar X
\approx
N\left(
\mu,
\frac{\sigma^2}{n}
\right).
$$

Therefore,

$$
\bar X
\approx
N\left(
0,
\frac{3/5}{15}
\right).
$$

Simplify the variance:

$$
\frac{3/5}{15}
=
\frac{3}{75}
=
\frac1{25}.
$$

Hence the standard error is

$$
\sigma_{\bar X}
=
\sqrt{\frac1{25}}
=
0.2.
$$

## Step 4 — Standardize

Lower z-score:

$$
z_L
=
\frac{-0.03-0}{0.2}
=
-0.15.
$$

Upper z-score:

$$
z_U
=
\frac{0.15-0}{0.2}
=
0.75.
$$

Thus,

$$
P(-0.03<\bar X<0.15)
\approx
\Phi(0.75)-\Phi(-0.15).
$$

Therefore,

$$
\boxed{
P(-0.03<\bar X<0.15)
\approx0.3329903400
}
$$

```python
# Question 6 verification

mu6 = 0
var6 = 3/5
n6 = 15
se6 = math.sqrt(var6/n6)

z6L = (-0.03-mu6)/se6
z6U = (0.15-mu6)/se6

p6 = norm.cdf(z6U) - norm.cdf(z6L)

print("Mean =", mu6)
print("Variance =", var6)
print("Standard error =", se6)
print("z lower =", z6L)
print("z upper =", z6U)
print("Probability =", p6)
```

### Why continuity correction matters

The binomial distribution assigns probability to integer values, while the normal distribution assigns probability to intervals. The $\pm0.5$ adjustment makes the continuous normal interval correspond more closely to the desired discrete binomial outcomes.

# Mini Theory Cheat Sheet — Questions 1–8

## Binomial PMF

$$
P(X=k)
=
\binom nkp^k(1-p)^{n-k}.
$$

## Independence

If $X$ and $Y$ are independent,

$$
P(A\cap B)=P(A)P(B),
$$

and for appropriate functions,

$$
E[g(X)h(Y)]
=
E[g(X)]E[h(Y)].
$$

## Normal Standardization

$$
Z=\frac{X-\mu}{\sigma}.
$$

## Second Moment

$$
E[X^2]
=
\operatorname{Var}(X)+E[X]^2.
$$

## Mean and Variance of a Sum

For independent variables,

$$
E[X+Y]=E[X]+E[Y]
$$

and

$$
\operatorname{Var}(X+Y)
=
\operatorname{Var}(X)+\operatorname{Var}(Y).
$$

## Sampling Distribution of the Mean

For a sample of size $n$,

$$
E[\bar X]=\mu
$$

and

$$
\operatorname{Var}(\bar X)
=
\frac{\sigma^2}{n}.
$$

If the population is normal, this distribution is exactly normal.

## Central Limit Theorem

For sufficiently large $n$,

$$
\bar X
\approx
N\left(
\mu,
\frac{\sigma^2}{n}
\right).
$$

For a sum

$$
Y=\sum_{i=1}^{n}X_i,
$$

$$
E[Y]=n\mu
$$

and

$$
\operatorname{Var}(Y)=n\sigma^2.
$$

## Continuity Correction

When approximating a discrete binomial probability with a continuous normal distribution, shift integer boundaries by $0.5$.

Example:

$$
P(Y=15)
$$

becomes approximately

$$
P(14.5<Y<15.5).
$$
