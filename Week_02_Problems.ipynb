
# Question 1 — Double integral over a rectangle

We need to calculate

$$
\iint_R (x+y^2)\,dA
$$

where

$$
0\le x\le1,\qquad 0\le y\le3.
$$

Because the region is rectangular, we can write

$$
\iint_R (x+y^2)\,dA
=
\int_0^1\int_0^3 (x+y^2)\,dy\,dx.
$$

Integrating first with respect to $y$:

$$
\int_0^3(x+y^2)\,dy
=
\left[xy+\frac{y^3}{3}\right]_0^3
=
3x+9.
$$

Now integrate with respect to $x$:

$$
\int_0^1(3x+9)\,dx
=
\left[\frac{3x^2}{2}+9x\right]_0^1
=
\frac{3}{2}+9.
$$

Therefore,

$$
\boxed{\iint_R (x+y^2)\,dA=\frac{21}{2}=10.5}
$$


````python

x, y = sp.symbols("x y", real=True)

q1 = sp.integrate(
    sp.integrate(x + y**2, (y, 0, 3)),
    (x, 0, 1)
)

q1
````


# Question 2 — Double integral over a triangular region

The triangle has vertices

$$
(0,0),\qquad (1,0),\qquad (1,1).
$$

The diagonal boundary is

$$
y=x.
$$

A convenient description of the region is

$$
0\le x\le1,\qquad 0\le y\le x.
$$

Therefore,

$$
\iint_R(2-y^2)\,dA
=
\int_0^1\int_0^x(2-y^2)\,dy\,dx.
$$

Integrating with respect to $y$:

$$
\int_0^x(2-y^2)\,dy
=
2x-\frac{x^3}{3}.
$$

Then

$$
\int_0^1\left(2x-\frac{x^3}{3}\right)\,dx
=
1-\frac{1}{12}.
$$

Hence,

$$
\boxed{\iint_R(2-y^2)\,dA=\frac{11}{12}}
$$


````python

q2 = sp.integrate(
    sp.integrate(2 - y**2, (y, 0, x)),
    (x, 0, 1)
)

q2
````


# Question 3 — Centroid of a triangle

The triangle has vertices

$$
(0,0),\qquad(0,-2),\qquad(7,4).
$$

For a region of uniform density, the centroid is

$$
\bar{x}
=
\frac{1}{A}\iint_R x\,dA,
\qquad
\bar{y}
=
\frac{1}{A}\iint_R y\,dA.
$$

## 3(a) Find the centroid using integrals

The upper boundary is the line joining $(0,0)$ to $(7,4)$:

$$
y=\frac{4}{7}x.
$$

The lower boundary joins $(0,-2)$ to $(7,4)$:

$$
y=\frac{6}{7}x-2.
$$

Thus,

$$
0\le x\le7,
$$

and

$$
\frac{6}{7}x-2
\le y\le
\frac{4}{7}x.
$$

### Area

$$
A
=
\int_0^7
\left(
\frac{4x}{7}
-
\left(\frac{6x}{7}-2\right)
\right)\,dx
=
7.
$$

### $x$-coordinate of the centroid

$$
\bar{x}
=
\frac{1}{7}
\int_0^7
x
\left(
2-\frac{2x}{7}
\right)\,dx
=
\frac{7}{3}.
$$

### $y$-coordinate of the centroid

$$
\bar{y}
=
\frac{1}{7}
\int_0^7
\int_{\frac{6x}{7}-2}^{\frac{4x}{7}}
y\,dy\,dx
=
\frac{2}{3}.
$$

Therefore,

$$
\boxed{
(\bar{x},\bar{y})
=
\left(\frac73,\frac23\right)
}
$$


````python

upper = 4*x/7
lower = 6*x/7 - 2

area = sp.integrate(upper - lower, (x, 0, 7))

x_moment = sp.integrate(
    x * (upper - lower),
    (x, 0, 7)
)

y_moment = sp.integrate(
    sp.integrate(y, (y, lower, upper)),
    (x, 0, 7)
)

x_bar = sp.simplify(x_moment / area)
y_bar = sp.simplify(y_moment / area)

area, x_bar, y_bar
````


## 3(b) Average the coordinates of the three vertices

Average the $x$-coordinates:

$$
\frac{0+0+7}{3}
=
\frac73.
$$

Average the $y$-coordinates:

$$
\frac{0+(-2)+4}{3}
=
\frac23.
$$

So the coordinate average is

$$
\boxed{
\left(\frac73,\frac23\right)
}
$$

which is exactly the same point found using integration.

### Why?

For a triangle with uniform density, the centroid is the intersection of its three medians, and its coordinates are exactly the arithmetic averages of the coordinates of its three vertices.

So, for any triangle with vertices

$$
(x_1,y_1),\quad(x_2,y_2),\quad(x_3,y_3),
$$

the centroid is

$$
\boxed{
\left(
\frac{x_1+x_2+x_3}{3},
\frac{y_1+y_2+y_3}{3}
\right)
}
$$



## 3(c) Centroid of the quadrilateral

The quadrilateral has vertices

$$
(0,0),\qquad(1,3),\qquad(3,6),\qquad(5,2).
$$

A key point is that **simply averaging the four vertices does not, in general, give the centroid of an arbitrary quadrilateral**.

We can use the result from part (b) by dividing the quadrilateral into two triangles.

Take the diagonal from $(0,0)$ to $(3,6)$.

### Triangle 1

Vertices:

$$
(0,0),\qquad(1,3),\qquad(3,6).
$$

Its area is

$$
A_1=\frac32,
$$

and, using the triangle rule from part (b), its centroid is

$$
C_1
=
\left(
\frac{0+1+3}{3},
\frac{0+3+6}{3}
\right)
=
\left(\frac43,3\right).
$$

### Triangle 2

Vertices:

$$
(0,0),\qquad(3,6),\qquad(5,2).
$$

Its area is

$$
A_2=12,
$$

and its centroid is

$$
C_2
=
\left(
\frac{0+3+5}{3},
\frac{0+6+2}{3}
\right)
=
\left(\frac83,\frac83\right).
$$

The quadrilateral centroid is the **area-weighted average** of the two triangle centroids:

$$
\bar{x}
=
\frac{
A_1\left(\frac43\right)
+
A_2\left(\frac83\right)
}{
A_1+A_2
}
=
\frac{68}{27},
$$

$$
\bar{y}
=
\frac{
A_1(3)
+
A_2\left(\frac83\right)
}{
A_1+A_2
}
=
\frac{73}{27}.
$$

Therefore,

$$
\boxed{
(\bar{x},\bar{y})
=
\left(
\frac{68}{27},
\frac{73}{27}
\right)
}
$$

Approximately,

$$
\boxed{
(\bar{x},\bar{y})
\approx
(2.519,\;2.704)
}
$$



# Question 4 — Jacobians

For a transformation

$$
x=x(u,v),
\qquad
y=y(u,v),
$$

the Jacobian is

$$
J(u,v)
=
\frac{\partial(x,y)}{\partial(u,v)}
=
\begin{vmatrix}
\frac{\partial x}{\partial u}
&
\frac{\partial x}{\partial v}
\\
\frac{\partial y}{\partial u}
&
\frac{\partial y}{\partial v}
\end{vmatrix}.
$$

## 4(a)

Given

$$
x=\frac{u}{v},
\qquad
y=v,
$$

the derivative matrix is

$$
\begin{pmatrix}
\frac1v & -\frac{u}{v^2}\\
0 & 1
\end{pmatrix}.
$$

Therefore,

$$
J(u,v)
=
\frac1v.
$$

Hence,

$$
\boxed{J(u,v)=\frac1v}
$$

provided $v\ne0$.

## 4(b)

Given

$$
x=\frac{u}{u+v},
\qquad
y=u+v,
$$

we have

$$
\frac{\partial x}{\partial u}
=
\frac{v}{(u+v)^2},
\qquad
\frac{\partial x}{\partial v}
=
-\frac{u}{(u+v)^2},
$$

and

$$
\frac{\partial y}{\partial u}=1,
\qquad
\frac{\partial y}{\partial v}=1.
$$

Therefore,

$$
J(u,v)
=
\frac{v}{(u+v)^2}
+
\frac{u}{(u+v)^2}
=
\frac1{u+v}.
$$

Thus,

$$
\boxed{J(u,v)=\frac1{u+v}}
$$

where $u+v\ne0$.


````python

u, v = sp.symbols("u v", nonzero=True)

x_a = u / v
y_a = v

J_a = sp.Matrix([
    [sp.diff(x_a, u), sp.diff(x_a, v)],
    [sp.diff(y_a, u), sp.diff(y_a, v)]
]).det()

x_b = u / (u + v)
y_b = u + v

J_b = sp.Matrix([
    [sp.diff(x_b, u), sp.diff(x_b, v)],
    [sp.diff(y_b, u), sp.diff(y_b, v)]
]).det()

sp.simplify(J_a), sp.simplify(J_b)
````


# Question 5 — Polar coordinates

The transformation is

$$
x=r\cos\theta,
\qquad
y=r\sin\theta.
$$

## 5(a) Jacobian

The Jacobian matrix is

$$
\begin{pmatrix}
\cos\theta & -r\sin\theta\\
\sin\theta & r\cos\theta
\end{pmatrix}.
$$

Its determinant is

$$
r\cos^2\theta+r\sin^2\theta
=
r.
$$

Therefore,

$$
\boxed{J(r,\theta)=r}
$$

and consequently

$$
\boxed{dA=r\,dr\,d\theta}.
$$

## 5(b) Integral over the unit disk

We need to calculate

$$
\int_C
\frac{1}{2\pi}
e^{-(x^2+y^2)/2}
\,dA
$$

over

$$
C:
x^2+y^2\le1.
$$

In polar coordinates,

$$
x^2+y^2=r^2,
$$

and the unit disk becomes

$$
0\le r\le1,
\qquad
0\le\theta\le2\pi.
$$

Therefore,

$$
\int_0^{2\pi}
\int_0^1
\frac{1}{2\pi}
e^{-r^2/2}
r
\,dr\,d\theta.
$$

The $\theta$ integral cancels the factor $1/(2\pi)$:

$$
\int_0^1 r e^{-r^2/2}\,dr.
$$

Using

$$
w=-\frac{r^2}{2},
\qquad
dw=-r\,dr,
$$

we obtain

$$
\int_0^1 r e^{-r^2/2}\,dr
=
1-e^{-1/2}.
$$

Hence,

$$
\boxed{
\int_C
\frac{1}{2\pi}
e^{-(x^2+y^2)/2}
\,dA
=
1-e^{-1/2}
}
$$

and numerically,

$$
\boxed{1-e^{-1/2}\approx0.393469}
$$



# Question 6 — Joint probability density and change of variables

We are given

$$
f(x,y)=2
$$

on

$$
R=\{(x,y):0<x<y<1\}.
$$

## 6(a) Verify the total probability

A convenient description is

$$
0<y<1,
\qquad
0<x<y.
$$

Therefore,

$$
\iint_R f(x,y)\,dA
=
\int_0^1\int_0^y 2\,dx\,dy.
$$

First,

$$
\int_0^y2\,dx=2y.
$$

Then,

$$
\int_0^1 2y\,dy=1.
$$

Thus,

$$
\boxed{\iint_R f(x,y)\,dA=1}
$$

which confirms that $f$ is normalized as a joint probability density.



## 6(b) Transform the region

The new variables are

$$
u=\frac{x}{y},
\qquad
v=y.
$$

Because

$$
0<x<y,
$$

dividing by the positive quantity $y$ gives

$$
0<\frac{x}{y}<1.
$$

Therefore,

$$
0<u<1.
$$

Since

$$
0<y<1
$$

and $v=y$,

$$
0<v<1.
$$

Thus the transformed region is simply

$$
\boxed{
S=\{(u,v):0<u<1,\;0<v<1\}
}
$$

which is the open unit square in the $uv$-plane.


````python

# Sketch of the transformed region S in the uv-plane
fig, ax = plt.subplots(figsize=(5, 5))

square = plt.Rectangle((0, 0), 1, 1, fill=False, linewidth=2)
ax.add_patch(square)

ax.set_xlim(-0.1, 1.1)
ax.set_ylim(-0.1, 1.1)
ax.set_xlabel("u")
ax.set_ylabel("v")
ax.set_title("Question 6(b): Region S in the uv-plane")
ax.set_aspect("equal")

plt.show()
````


## 6(c) Why is $S$ preferred?

The original region $R$ is triangular and has a variable boundary $x<y$.

The transformed region $S$ is a rectangle:

$$
0<u<1,
\qquad
0<v<1.
$$

Therefore the integration limits become independent and much simpler.

## 6(d) Jacobian

From

$$
u=\frac{x}{y},
\qquad
v=y,
$$

the inverse transformation is

$$
x=uv,
\qquad
y=v.
$$

The Jacobian needed for change of variables is

$$
\frac{\partial(x,y)}{\partial(u,v)}
=
\begin{vmatrix}
v & u\\
0 & 1
\end{vmatrix}
=
v.
$$

Therefore,

$$
\boxed{
\left|
\frac{\partial(x,y)}{\partial(u,v)}
\right|
=v
}
$$

and

$$
\boxed{dx\,dy=v\,du\,dv}.
$$

For completeness, the forward Jacobian is

$$
\frac{\partial(u,v)}{\partial(x,y)}
=
\frac1y
=
\frac1v.
$$

As a check, the probability integral becomes

$$
\int_0^1\int_0^1 2v\,du\,dv=1.
$$

# 1. Double Integrals

A double integral

$$
\iint_R f(x,y)\,dA
$$

adds the values of a function over a two-dimensional region $R$.

You can imagine the region divided into tiny pieces of area $\Delta A$. Each contributes approximately

$$
f(x,y)\Delta A.
$$

Adding all pieces and taking the limit gives the double integral.

## Rectangle

If

$$
a\le x\le b,
\qquad
c\le y\le d,
$$

then

$$
\iint_R f(x,y)\,dA
=
\int_a^b\int_c^d f(x,y)\,dy\,dx.
$$

So:

$$
\boxed{\text{Rectangle} \Rightarrow \text{constant limits}}
$$



# 2. Non-Rectangular Regions

For triangular or curved regions, one limit often depends on the other variable.

Example:

$$
0\le x\le1,
\qquad
0\le y\le x.
$$

Then

$$
\iint_R f(x,y)\,dA
=
\int_0^1\int_0^x f(x,y)\,dy\,dx.
$$

## How to choose limits

Ask:

1. What is the outer variable?
2. What is its full range?
3. For each fixed outer-variable value, where does the inner variable start?
4. Where does it end?

A good rule is:

$$
\boxed{\text{Sketch the region before choosing limits}}
$$



# 3. Area, Average Value, and Centroid

If the integrand is $1$,

$$
\boxed{
A=\iint_R dA
}
$$

gives the area.

The average value of a function over a region is

$$
\boxed{
f_{\text{avg}}
=
\frac{1}{A}
\iint_R f(x,y)\,dA
}
$$

For a uniform lamina, the centroid is

$$
\boxed{
\bar{x}
=
\frac{1}{A}
\iint_R x\,dA
}
$$

$$
\boxed{
\bar{y}
=
\frac{1}{A}
\iint_R y\,dA
}
$$

Think of the centroid as the balance point of a thin sheet.

## Triangle shortcut

For vertices

$$
(x_1,y_1),\quad
(x_2,y_2),\quad
(x_3,y_3),
$$

the centroid is

$$
\boxed{
\left(
\frac{x_1+x_2+x_3}{3},
\frac{y_1+y_2+y_3}{3}
\right)
}
$$

This shortcut is special to triangles.

For a general quadrilateral, simply averaging the four vertices is not generally correct. A safe method is to split it into triangles and use an area-weighted average of their centroids.



# 4. Jacobian

Suppose

$$
x=x(u,v),
\qquad
y=y(u,v).
$$

The Jacobian is

$$
\boxed{
J(u,v)
=
\frac{\partial(x,y)}{\partial(u,v)}
}
$$

with

$$
J(u,v)
=
\begin{vmatrix}
\frac{\partial x}{\partial u}
&
\frac{\partial x}{\partial v}
\\
\frac{\partial y}{\partial u}
&
\frac{\partial y}{\partial v}
\end{vmatrix}.
$$

So,

$$
\boxed{
J
=
x_u y_v-x_v y_u
}
$$

## Meaning

The Jacobian tells us how much a transformation locally stretches or compresses area.

If a tiny area in the $uv$-plane is

$$
du\,dv,
$$

then in the $xy$-plane it becomes approximately

$$
\boxed{
dA
=
|J|\,du\,dv
}
$$

Therefore:

$$
\boxed{\text{Jacobian = local area-scaling factor}}
$$



# 5. Change of Variables

Change of variables is useful when the original region or integrand is awkward.

The general formula is

$$
\boxed{
\iint_R f(x,y)\,dx\,dy
=
\iint_S
f(x(u,v),y(u,v))
\left|
\frac{\partial(x,y)}{\partial(u,v)}
\right|
du\,dv
}
$$

## Procedure

1. Define the new variables.
2. Find the inverse transformation.
3. Transform the region.
4. Compute the Jacobian.
5. Transform the integrand.
6. Integrate over the new region.

In Question 6,

$$
0<x<y<1
$$

becomes

$$
0<u<1,
\qquad
0<v<1.
$$

That turns a triangular region into a rectangle, which is easier to integrate over.



# 6. Polar Coordinates

Polar coordinates use distance $r$ and angle $\theta$:

$$
\boxed{x=r\cos\theta}
$$

$$
\boxed{y=r\sin\theta}
$$

and

$$
\boxed{x^2+y^2=r^2}
$$

The Jacobian is

$$
\boxed{
\frac{\partial(x,y)}{\partial(r,\theta)}=r
}
$$

so

$$
\boxed{
dA=r\,dr\,d\theta
}
$$

## When to use polar coordinates

Think polar coordinates when you see:

- disks;
- circles;
- annuli;
- expressions involving $x^2+y^2$.

So:

$$
\boxed{\text{Circle} \Rightarrow \text{Polar coordinates}}
$$

### Common mistake

Do not forget the factor $r$ in

$$
dA=r\,dr\,d\theta.
$$

A matrix has rows and columns.

has 2 rows and 2 columns.

So:

$$
\boxed{\text{Addition/subtraction requires the same shape}}

# 9. Matrix Multiplication and Transpose

If

$$
A
$$

is $m\times n$ and

$$
B
$$

is $n\times p$, then

$$
AB
$$

exists and has dimensions

$$
m\times p.
$$

So:

$$
\boxed{
(m\times n)(n\times p)\Rightarrow m\times p
}
$$

Matrix multiplication is generally not commutative:

$$
\boxed{AB\ne BA}

The transpose swaps rows and columns:

$$
A=
\begin{pmatrix}

c&d
\end{pmatrix}
\Rightarrow

\begin{pmatrix}

\end{pmatrix}.
$$

Useful identity:

$$
\boxed{
(AB)^T=B^TA^T
}

# 10. Determinant and Inverse

For

$$

\begin{pmatrix}

c&d
\end{pmatrix},

the determinant is

$$
\boxed{
\det(A)=ad-bc
}

The determinant tells us whether a square matrix is invertible.

\det(A)\ne0

# 12. Gauss–Jordan Elimination

A system of equations can be represented as an augmented matrix.

Example:

$$
4x-2y=32
$$

$$
x-3y=3
$$

becomes

$$
\left(
\begin{array}{cc|c}
4&-2&32\\
1&-3&3
\end{array}
\right).
$$

## Allowed elementary row operations

### Swap rows

$$
R_i\leftrightarrow R_j
$$

### Multiply a row by a nonzero constant

$$
R_i\leftarrow cR_i
$$

### Add a multiple of one row to another

$$
R_i\leftarrow R_i+cR_j
$$

The goal is RREF.

For a unique two-variable solution, the final form looks like

$$
\left(
\begin{array}{cc|c}
1&0&a\\
0&1&b
\end{array}
\right),
$$

so

$$
x=a,\qquad y=b.

# 15. Eigenvalues and Eigenvectors

An eigenvector is a nonzero vector whose direction is preserved by a matrix transformation.

$$
\boxed{
A\mathbf v=\lambda\mathbf v
}
$$

where:

- $\mathbf v$ is an eigenvector;
- $\lambda$ is its eigenvalue.

## Interpretation

If

$$
\lambda=2,
$$

the vector keeps its direction and doubles in length.

If

$$
\lambda=-4,
$$

the vector becomes four times longer and reverses direction.

If

$$
\lambda=1,
$$

the vector is unchanged.

## How to find an eigenvector for a known eigenvalue

Start with

$$
A\mathbf v=\lambda\mathbf v.
$$

Then

$$
\boxed{
(A-\lambda I)\mathbf v=0
}
$$

Solve this homogeneous system for a nonzero vector.

A nonzero solution exists only when

$$
\boxed{
\det(A-\lambda I)=0
}
$$

Remember:

$$
\boxed{\mathbf v\ne\mathbf0}
$$

because the zero vector is never an eigenvector.

\left(
\frac{x_1+x_2+x_3}{3},
\frac{y_1+y_2+y_3}{3}
\right)
}
$$

## Jacobian

$$
\boxed{
\frac{\partial(x,y)}{\partial(u,v)}
=
\begin{vmatrix}
x_u&x_v\\
y_u&y_v
\end{vmatrix}
}
$$

## Change of variables

$$
\boxed{
dx\,dy
=
\left|
\frac{\partial(x,y)}{\partial(u,v)}
\right|
du\,dv
}
$$

## Polar coordinates

$$
\boxed{x=r\cos\theta}
$$

$$
\boxed{y=r\sin\theta}
$$

$$
\boxed{x^2+y^2=r^2}
$$

$$
\boxed{dA=r\,dr\,d\theta}
$$

## Probability density

$$
\boxed{
\iint_R f(x,y)\,dA=1
}
$$

## Determinant

$$
\boxed{
\det
\begin{pmatrix}
a&b\\
c&d
\end{pmatrix}
=
ad-bc
}
$$

## Inverse

$$
\boxed{
A^{-1}
=
\frac1{ad-bc}
\begin{pmatrix}
d&-b\\
-c&a
\end{pmatrix}
}
$$

## Vector norm

$$
\boxed{
\|\mathbf v\|
=
\sqrt{v_1^2+\cdots+v_n^2}
}
$$

## Rotation matrix

$$
\boxed{
R_\theta
=
\begin{pmatrix}
\cos\theta&-\sin\theta\\
\sin\theta&\cos\theta
\end{pmatrix}
}
$$

## Eigenvector equation

$$
\boxed{
A\mathbf v=\lambda\mathbf v
}
$$

or

$$
\boxed{
(A-\lambda I)\mathbf v=0
}
$$



# 20. Final Mental Model

Remember the assignment through two pictures.

## Multivariable Calculus

Think of a surface and a region underneath it.

You want to:

- accumulate values;
- measure area;
- locate balance points;
- simplify difficult regions;
- use coordinates adapted to circles.

So the progression is

$$
\boxed{
\text{Double Integrals}
\rightarrow
\text{Centroids}
\rightarrow
\text{Jacobians}
\rightarrow
\text{Change of Variables}
\rightarrow
\text{Polar Coordinates}
}
$$

## Linear Algebra

Think of vectors as arrows and matrices as machines.

You want to:

- combine matrices;
- solve systems;
- understand geometric transformations;
- rotate vectors;
- identify special directions preserved by transformations.

So the progression is

$$
\boxed{
\text{Matrices}
\rightarrow
\text{Systems}
\rightarrow
\text{Transformations}
\rightarrow
\text{Rotations}
\rightarrow
\text{Eigenvectors}
}
$$

If these two mental pictures are clear, the formulas become much easier to remember.
