# Question 1 — Clustering vs Classification

## Professor's question

**In your own words, what is the difference between clustering and classification?**

## Answer

The main difference is whether the correct group labels are already known during learning.

### Clustering

**Clustering is an unsupervised-learning task.**

We start with data points but we do **not** provide known class labels. The algorithm tries to discover natural groups based on similarity.

For example, suppose we have customers described by:

$$
(\text{annual spending},\text{purchase frequency},\text{average order value}).
$$

A clustering algorithm might discover groups such as:

- low-frequency / high-value customers;
- frequent / medium-value customers;
- infrequent / low-value customers.

The groups are discovered from the structure of the data rather than supplied in advance.

### Classification

**Classification is a supervised-learning task.**

We train with examples whose correct classes are already known. The model learns a rule that maps input features to one of the predefined classes.

For example, an email-classification system might learn:

$$
\text{email features}
\longrightarrow
\{\text{legitimate},\text{spam},\text{phishing}\}.
$$

During training, examples are already labeled with their correct category.

---

## Core distinction

$$
\boxed{
\text{Clustering: discover groups without known labels}
}
$$

$$
\boxed{
\text{Classification: predict predefined labels from labeled examples}
}
$$

A simple way to remember the difference is:

> **Clustering asks, “What groups seem to exist?”**  
> **Classification asks, “Which known group does this example belong to?”**

# Question 5 — Machine-Learning Foundations

## 5(a) Where is minimization used in machine learning?

Minimization is used when training a model.

A machine-learning model usually has a **loss function** or **error function** that measures how wrong the model's predictions are.

For example,

$$
L(\theta)
$$

can represent the loss as a function of model parameters

$$
\theta.
$$

Training attempts to find parameters that make the loss as small as possible:

$$
\boxed{
\theta^*
=
\arg\min_{\theta}L(\theta)
}
$$

In a neural network, the parameters include the weights and biases.

For example,

$$
\theta
=
(w_1,w_2,\ldots,b_1,b_2,\ldots).
$$

Gradient-based optimization repeatedly adjusts these parameters to reduce the loss.

So minimization appears directly in:

- fitting linear and logistic models;
- training neural networks;
- centroid optimization in k-means;
- minimizing prediction error;
- selecting model parameters.

---

## 5(b) Gradient descent vs stochastic gradient descent

### Batch gradient descent

In ordinary batch gradient descent, we calculate the gradient using the **entire training dataset** before making one parameter update.

If the total loss is

$$
L
=
\sum_{i=1}^{N}L_i,
$$

then one update has the form

$$
\theta
:=
\theta
-
\eta\nabla L(\theta).
$$

The entire dataset contributes to the gradient.

### Stochastic gradient descent

In stochastic gradient descent (SGD), we update the model using **one training example at a time**.

For sample $i$,

$$
\theta
:=
\theta
-
\eta\nabla L_i(\theta).
$$

So SGD performs many smaller and more frequent updates.

### Practical difference

Batch gradient descent:

- uses all examples for each update;
- gives a smoother, deterministic gradient for a fixed dataset;
- can be computationally expensive on large datasets.

SGD:

- updates after individual examples;
- is computationally cheaper per update;
- produces noisier parameter movement;
- can learn efficiently from large datasets.

A **mini-batch** method lies between these two extremes and is common in modern neural-network training.

---

## 5(c) What are hyperparameters?

Hyperparameters are settings that we choose for the learning procedure or model structure rather than values directly learned as model parameters.

Examples include:

- learning rate $\eta$;
- number of hidden layers;
- number of neurons per hidden layer;
- number of epochs;
- batch size;
- activation-function choice;
- regularization strength;
- optimizer choice.

For example,

$$
\eta=0.01
$$

is a hyperparameter.

By contrast, neural-network weights such as

$$
w_1,w_2
$$

and biases such as

$$
b
$$

are **parameters**, because training changes them based on data.

---

## 5(d) What is a perceptron?

A perceptron is a simple artificial neuron.

It takes inputs

$$
x_1,x_2,\ldots,x_n,
$$

multiplies them by weights,

$$
w_1,w_2,\ldots,w_n,
$$

adds a bias,

$$
b,
$$

and calculates

$$
s
=
w_1x_1+\cdots+w_nx_n+b.
$$

An activation or threshold rule then converts this weighted sum into an output.

The basic structure is therefore

$$
\boxed{
\text{inputs}
\rightarrow
\text{weighted sum + bias}
\rightarrow
\text{activation}
\rightarrow
\text{output}
}
$$

A perceptron is one of the fundamental building blocks that motivated modern neural networks.

## 5(e) Two examples where matrices are used in machine learning

### Example 1 — Data matrices

A dataset containing $N$ observations and $d$ features can be represented as a matrix

$$
X
\in
\mathbb{R}^{N\times d}.
$$

Each row can represent one observation, while each column represents a feature.

For example,

$$
X=
\begin{bmatrix}
x_{11}&x_{12}&x_{13}\\
x_{21}&x_{22}&x_{23}\\
x_{31}&x_{32}&x_{33}
\end{bmatrix}.
$$

### Example 2 — Neural-network weight matrices

Suppose one layer has $n$ inputs and the next has $m$ neurons.

The connections can be represented by a weight matrix

$$
W
\in
\mathbb{R}^{m\times n}.
$$

A layer calculation can then be written compactly as

$$
\mathbf{s}
=
W\mathbf{x}+\mathbf{b}.
$$

So matrices efficiently represent both the data and the transformations applied by a neural network.

## 5(f) Why are probability and statistics important in machine learning?

Machine learning deals with data that contain uncertainty, noise, variability, and incomplete information.

Probability and statistics provide the mathematical tools needed to:

- model uncertainty;
- describe data distributions;
- estimate unknown parameters;
- calculate expected values and variances;
- evaluate model performance;
- construct confidence intervals;
- perform hypothesis tests;
- reason about prediction probabilities;
- measure relationships between variables;
- understand sampling variability.

For example, a classifier might output

$$
P(Y=\text{phishing}\mid X)=0.92.
$$

Probability gives meaning to this uncertainty.

Statistics then helps us determine whether observed performance is likely to generalize beyond the training sample.

Thus,

$$
\boxed{
\text{probability models uncertainty}
}
$$

while

$$
\boxed{

$$
E
=
\frac12
\sum_{i=1}^{4}

E_i

$$
\frac{\partial \hat y_i}{\partial s_i}

$$

## Question 6(d) — 100 epochs of stochastic gradient descent

The professor specifies:

$$
\eta=1
$$

and

$$
100\text{ epochs}.
$$

The training examples are processed in the fixed order

$$
i=1,2,3,4
$$

during every epoch.

Because the method is **stochastic gradient descent**, we update after **each individual input**, rather than calculating one full-dataset gradient and updating only once per epoch.

---

## Per-example loss

For one example,

$$
E_i
=
\frac12(\hat y_i-y_i)^2.
$$

For the current example, define

$$
\delta_i
=
(\hat y_i-y_i)\hat y_i(1-\hat y_i).
$$

Then

$$
\frac{\partial E_i}{\partial w_1}
=
\delta_i x_{1i},
$$

$$
\frac{\partial E_i}{\partial w_2}
=
\delta_i x_{2i},
$$

and

$$
\frac{\partial E_i}{\partial b}
=
\delta_i.
$$

The professor's SGD updates are therefore

$$
w_1
:=
w_1-\eta\delta_i x_{1i},
$$

$$
w_2
:=
w_2-\eta\delta_i x_{2i},
$$

$$
b
:=
b-\eta\delta_i.
$$

For this problem,

$$
\eta=1.
$$

A crucial implementation detail is that all three gradients for one example are calculated using the **same current forward pass** before the next example is processed.

---

## Python implementation

The code below performs exactly:

$$
100
\times
4
=
400
$$

individual SGD updates.

```python
# Question 6(d) — professor-specified SGD

def total_error(w1, w2, b):
    predictions = []
    for x1, x2 in X:
        s = w1*x1 + w2*x2 + b
        predictions.append(sigmoid(s))
    predictions = np.array(predictions)
    return 0.5*np.sum((predictions-y)**2)

w1_train = 0.4
w2_train = -0.2
b_train = 0.1

eta = 1.0
epoch_errors = []

for epoch in range(100):

    # Professor specifies the fixed order i = 1,2,3,4.
    for (x1, x2), yi in zip(X, y):

        # Forward pass using the current parameters.
        s = w1_train*x1 + w2_train*x2 + b_train
        yhat = sigmoid(s)

        # Derivative common to all three parameters for this example.
        delta_i = (yhat-yi)*yhat*(1-yhat)

        # SGD parameter updates.
        w1_train = w1_train - eta*delta_i*x1
        w2_train = w2_train - eta*delta_i*x2
        b_train  = b_train  - eta*delta_i

    epoch_errors.append(total_error(w1_train, w2_train, b_train))

print("After 100 epochs:")
print("w1 =", w1_train)
print("w2 =", w2_train)
print("b  =", b_train)
print("total error =", epoch_errors[-1])
```

# Week 08 Mature Cheat Sheet

## 1. Clustering

Clustering is an **unsupervised-learning** problem.

Input:

$$
\mathbf{x}_1,\ldots,\mathbf{x}_N
$$

without known labels.

Goal:

$$
\boxed{
\text{discover groups of similar observations}
}
$$

Example:

$$
\text{k-means clustering}.
$$

---

## 2. Classification

Classification is a **supervised-learning** problem.

Training data have the form

$$
(\mathbf{x}_i,y_i),
$$

where $y_i$ is the known class.

Goal:

$$
\boxed{
\mathbf{x}
\longrightarrow
\text{predefined class}
}
$$

---

## 3. Neural-network input and output layers

### Input layer

The number of input neurons normally corresponds to the number of input features:

$$
\boxed{
n_{\text{input}}
=
n_{\text{features}}
}
$$

### Output layer

For a multiclass design following the Week 08 framing:

$$
\boxed{
n_{\text{output}}
=
n_{\text{classes}}
}
$$

---

## 4. Parameters vs hyperparameters

### Parameters

Learned from data:

$$
w_1,w_2,\ldots,b_1,b_2,\ldots
$$

### Hyperparameters

Chosen by the practitioner:

- learning rate;
- number of layers;
- number of neurons;
- epochs;
- batch size;
- activation functions;
- regularization.

---

## 5. Perceptron / artificial neuron

The weighted sum is

$$
s
=
\sum_jw_jx_j+b.
$$

An activation function converts $s$ into an output.

---

## 6. Sigmoid function

$$
\boxed{
S(s)
=
\frac1{1+e^{-s}}
}
$$

Range:

$$
0<S(s)<1.
$$

Derivative:

$$
\boxed{
S'(s)
=
S(s)(1-S(s))
}
$$

If

$$
\hat y=S(s),
$$

then

$$
\boxed{
S'(s)
=
\hat y(1-\hat y)
}
$$

This identity is essential for Question 6.

---

## 7. Squared-error loss

For one observation:

$$
\boxed{
E_i
=
\frac12(\hat y_i-y_i)^2
}
$$

The factor $\frac12$ makes the derivative convenient:

$$
\frac{\partial E_i}{\partial\hat y_i}
=
\hat y_i-y_i.
$$

For the Week 08 OR example:

$$
\boxed{
E
=
\frac12
\sum_i
(\hat y_i-y_i)^2
}
$$

---

## 8. Chain rule in a neural network

If

$$
E_i
\rightarrow
\hat y_i
\rightarrow
s_i
\rightarrow
w,
$$

then

$$
\frac{\partial E_i}{\partial w}
=
\frac{\partial E_i}{\partial\hat y_i}
\frac{\partial\hat y_i}{\partial s_i}
\frac{\partial s_i}{\partial w}.
$$

For the sigmoid neuron:

$$
\boxed{
\frac{\partial E_i}{\partial w_j}
=
(\hat y_i-y_i)
\hat y_i(1-\hat y_i)
x_{ji}
}
$$

Bias derivative:

$$
\boxed{
\frac{\partial E_i}{\partial b}
=
(\hat y_i-y_i)
\hat y_i(1-\hat y_i)
}
$$

---

## 9. Gradient descent

General update:

$$
\boxed{
\theta
:=
\theta
-
\eta
\frac{\partial E}{\partial\theta}
}
$$

where

$$
\eta
$$

is the learning rate.

The negative sign means we move **against** the gradient to reduce the error.

---

## 10. Stochastic gradient descent

SGD updates the parameters after individual examples:

$$
\theta
:=
\theta
-
\eta
\nabla E_i.
$$

One epoch means one pass through all training examples.

In Question 6:

$$
4\text{ examples}
\times
100\text{ epochs}
=
400\text{ SGD updates}.
$$

---

## 11. Why the bias matters

Without a bias, the decision boundary is forced through the origin.

With

$$
s=w_1x_1+w_2x_2+b,
$$

the bias lets the model shift the boundary.

For the OR gate, training makes the bias negative while the two weights become strongly positive.

That causes

$$
(0,0)
$$

to produce a low output while any input containing a 1 produces a high output.

---

## 12. Matrix view of a neural-network layer

A layer can be written as

$$
\boxed{
\mathbf{s}
=
W\mathbf{x}
+
\mathbf{b}
}
$$

where:

- $\mathbf{x}$ is the input vector;
- $W$ is the weight matrix;
- $\mathbf{b}$ is the bias vector.

This is why linear algebra is central to machine learning.

# Common Week 08 Mistakes to Avoid

1. **Do not confuse clustering and classification.**  
   Clustering has no provided target labels; classification does.

2. **Do not call weights and biases hyperparameters.**  
   They are trainable parameters.

3. **Do not forget the bias term.**

$$
s=w_1x_1+w_2x_2+b.
$$

4. **Do not forget the sigmoid derivative.**

$$
S'(s)=S(s)(1-S(s)).
$$

5. **Do not omit the input value when differentiating a weight.**

$$
\frac{\partial s}{\partial w_j}=x_j.
$$

6. **Do not treat SGD as one update per epoch.**  
   In this assignment the model is updated after every training example.

7. **Do not randomize the input order for Question 6.**  
   The professor explicitly asks us to loop through inputs $i=1$ to $4$ repeatedly.

8. **Do not round the sigmoid output during training.**  
   Training uses the continuous $\hat y$ value. Rounding is only used afterward to interpret the class.

```python
# Recalculate initial values independently.

def sig_check(s):
    return 1/(1+math.exp(-s))

expected_initial = np.array([
    0.52497918747894,
    0.6224593312018546,
    0.47502081252106,
    0.574442516811659
])

actual_initial = []

for x1, x2 in X:
    actual_initial.append(sig_check(0.4*x1 - 0.2*x2 + 0.1))

actual_initial = np.array(actual_initial)

assert np.allclose(actual_initial, expected_initial, atol=1e-12)

expected_error = 0.43742121133312106
actual_error = 0.5*np.sum((actual_initial-y)**2)

assert abs(actual_error-expected_error) < 1e-12

delta_check = (actual_initial-y)*actual_initial*(1-actual_initial)

assert abs(np.sum(delta_check*X[:,0]) - (-0.19275452254222167)) < 1e-12
assert abs(np.sum(delta_check*X[:,1]) - (-0.2349482948247647)) < 1e-12
assert abs(np.sum(delta_check) - (-0.19275452254222167)) < 1e-12

# Re-run professor-specified SGD independently.

wc1, wc2, bc = 0.4, -0.2, 0.1

for _ in range(100):
    for (x1, x2), yi in zip(X, y):
        s = wc1*x1 + wc2*x2 + bc
        yh = sig_check(s)
        d = (yh-yi)*yh*(1-yh)

        wc1 -= d*x1
        wc2 -= d*x2
        bc  -= d

assert abs(wc1 - 3.3698199104597046) < 1e-12
assert abs(wc2 - 3.3392144298733846) < 1e-12
assert abs(bc  - (-1.3539152310516525)) < 1e-12

expected_final = np.array([
    0.2052310165282785,
    0.8824568797895949,
    0.8792449271323881,
    0.9952983035728866
])

final_check = []

for x1, x2 in X:
    s = wc1*x1 + wc2*x2 + bc
    final_check.append(sig_check(s))

final_check = np.array(final_check)

assert np.allclose(final_check, expected_final, atol=1e-12)

assert np.array_equal((final_check >= 0.5).astype(int),
                      np.array([0,1,1,1]))

print("All Week 08 Question 6 numerical checks passed.")
```
