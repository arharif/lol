# Question 1 — Independent Binomial Random Variables

The assignment gives

$$
X_1\sim b\left(3,\frac12\right)
$$

and

$$
X_2\sim b\left(5,\frac12\right),
$$

and states that $X_1$ and $X_2$ are independent.

For a binomial random variable,

$$
P(X=k)
=
\binom{n}{k}p^k(1-p)^{n-k}.
$$

Because $p=1/2$ here,

$$
P(X=k)
=
\binom{n}{k}\left(\frac12\right)^n.
$$

## 1(a) Find $P(X_1=2,\;X_2=4)$

Because $X_1$ and $X_2$ are independent,

$$
P(X_1=2,\;X_2=4)
=
P(X_1=2)P(X_2=4).
$$

First,

$$
P(X_1=2)
=
\binom32
\left(\frac12\right)^3
=
3\cdot\frac18
=
\frac38.
$$

Next,

$$
P(X_2=4)
=
\binom54
\left(\frac12\right)^5
=
5\cdot\frac1{32}
=
\frac5{32}.
$$

Therefore,

$$
P(X_1=2,\;X_2=4)
=
\frac38\cdot\frac5{32}
=
\frac{15}{256}.
$$

So,

$$
\boxed{
P(X_1=2,\;X_2=4)
=
\frac{15}{256}
\approx0.05859375
}
$$

## 1(b) Find $P(X_1+X_2=7)$

Since the variables are independent and have the same success probability $p=1/2$,

$$
X_1+X_2
\sim
b\left(3+5,\frac12\right)
=
b\left(8,\frac12\right).
$$

Therefore,

$$
P(X_1+X_2=7)
=
\binom87
\left(\frac12\right)^8.
$$

Thus,

$$
=
8\cdot\frac1{256}
=
\frac8{256}
=
\frac1{32}.
$$

Hence,

$$
\boxed{
P(X_1+X_2=7)
=
\frac1{32}
=
0.03125
}
$$

### Direct check

The only possibilities are

$$
(X_1,X_2)=(2,5)
$$

or

$$
(X_1,X_2)=(3,4).
$$

Their probabilities add to the same result.

# Question 4 — Mean and Variance of a Sum

The assignment gives a random sample of size 2 from the pdf

$$
f(x)=6x(1-x),
\qquad
0<x<1.
$$

Let

$$
Y=X_1+X_2.
$$

Because $X_1$ and $X_2$ form a random sample, they are independent and identically distributed.

We first find the mean and variance of one observation $X$.

## Step 1 — Find $E[X]$

$$
E[X]
=
\int_0^1x\,6x(1-x)\,dx.
$$

Expand:

$$
E[X]
=
6\int_0^1(x^2-x^3)\,dx.
$$

Therefore,

$$
=
6\left(
\frac13-\frac14
\right)
=
6\left(\frac1{12}\right)
=
\frac12.
$$

Thus,

$$
E[X]=\frac12.
$$

## Step 2 — Find $E[X^2]$

$$
E[X^2]
=
\int_0^1x^2\,6x(1-x)\,dx.
$$

Therefore,

$$
=
6\int_0^1(x^3-x^4)\,dx.
$$

So,

$$
=
6\left(
\frac14-\frac15
\right)
=
6\left(\frac1{20}\right)
=
\frac3{10}.
$$

## Step 3 — Find $\operatorname{Var}(X)$

$$
\operatorname{Var}(X)
=
E[X^2]-E[X]^2.
$$

Thus,

$$
=
\frac3{10}
-
\left(\frac12\right)^2
=
\frac3{10}-\frac14.
$$

Using denominator 20,

$$
=
\frac6{20}-\frac5{20}
=
\frac1{20}.
$$

## Step 4 — Find the mean of $Y$

$$
E[Y]
=
E[X_1+X_2].
$$

By linearity,

$$
E[Y]
=
E[X_1]+E[X_2]
=
\frac12+\frac12
=
1.
$$

Therefore,

$$
\boxed{E[Y]=1}.
$$

## Step 5 — Find the variance of $Y$

Since $X_1$ and $X_2$ are independent,

$$
\operatorname{Var}(Y)
=
\operatorname{Var}(X_1)+\operatorname{Var}(X_2).
$$

Thus,

$$
=
\frac1{20}+\frac1{20}
=
\frac1{10}.
$$

Therefore,

$$
\boxed{
\operatorname{Var}(Y)=\frac1{10}=0.1
}
$$

# Question 6 — Central Limit Theorem for a Sample Mean

The professor explicitly says to use the **Central Limit Theorem** for Questions 6–8.

The population pdf is

$$
f(x)
=
\frac32x^2,
\qquad
-1<x<1.
$$

A random sample of size

$$
n=15
$$

is taken, and $\bar X$ denotes the sample mean.

We want

$$
P(-0.03<\bar X<0.15).
$$

To use the CLT, first find the population mean and variance.

## Step 1 — Population mean

$$
\mu
=
E[X]
=
\int_{-1}^{1}
x\left(\frac32x^2\right)dx.
$$

Thus,

$$
\mu
=
\frac32
\int_{-1}^{1}x^3\,dx.
$$

Because $x^3$ is an odd function integrated symmetrically over $[-1,1]$,

$$
\boxed{\mu=0}.
$$

## Step 2 — Population variance

Since $\mu=0$,

$$
\operatorname{Var}(X)=E[X^2].
$$

Now,

$$
E[X^2]
=
\int_{-1}^{1}
x^2\left(\frac32x^2\right)dx.
$$

Therefore,

$$
=
\frac32
\int_{-1}^{1}x^4\,dx.
$$

Since $x^4$ is even,

$$
=
\frac32
\left(
2\int_0^1x^4\,dx
\right).
$$

Thus,

$$
=
\frac32
\left(
2\cdot\frac15
\right)
=
\frac35.
$$

Therefore,

$$
\boxed{
\sigma^2=\frac35
}.
$$

## Step 3 — Sampling distribution of $\bar X$

By the Central Limit Theorem,

$$
\bar X
\approx
N\left(
\mu,
\frac{\sigma^2}{n}
\right).
$$

Therefore,

$$
\bar X
\approx
N\left(
0,
\frac{3/5}{15}
\right).
$$

Simplify the variance:

$$
\frac{3/5}{15}
=
\frac{3}{75}
=
\frac1{25}.
$$

Hence the standard error is

$$
\sigma_{\bar X}
=
\sqrt{\frac1{25}}
=
0.2.
$$

## Step 4 — Standardize

Lower z-score:

$$
z_L
=
\frac{-0.03-0}{0.2}
=
-0.15.
$$

Upper z-score:

$$
z_U
=
\frac{0.15-0}{0.2}
=
0.75.
$$

Thus,

$$
P(-0.03<\bar X<0.15)
\approx
\Phi(0.75)-\Phi(-0.15).
$$

Therefore,

$$
\boxed{
P(-0.03<\bar X<0.15)
\approx0.3329903400
}
$$

````python
# Question 6 verification

mu6 = 0
var6 = 3/5
n6 = 15
se6 = math.sqrt(var6/n6)

z6L = (-0.03-mu6)/se6
z6U = (0.15-mu6)/se6

p6 = norm.cdf(z6U) - norm.cdf(z6L)

print("Mean =", mu6)
print("Variance =", var6)
print("Standard error =", se6)
print("z lower =", z6L)
print("z upper =", z6U)
print("Probability =", p6)
````

# Question 9 — Type I and Type II Errors

The problem compares two possible bowls.

For Bowl A,

$$
p=\frac{100}{300}=\frac13.
$$

For Bowl B,

$$
p=\frac{200}{300}=\frac23.
$$

The hypotheses are therefore

$$
H_0:p=\frac13
$$

against

$$
H_1:p=\frac23.
$$

Three balls are drawn **with replacement**, so under either hypothesis,

$$
X\sim\operatorname{Binomial}(3,p).
$$

The critical region is

$$
X\ge2.
$$

That means we reject $H_0$ whenever we observe 2 or 3 red balls.

---

## Type I error and $\alpha$

A Type I error means:

> Reject $H_0$ even though $H_0$ is true.

Therefore,

$$
\alpha
=
P(\text{reject }H_0\mid H_0\text{ true}).
$$

Under $H_0$,

$$
p=\frac13.
$$

Since the rejection region is $X\ge2$,

$$
\alpha
=
P_{p=1/3}(X\ge2).
$$

Thus,

$$
\alpha
=
P(X=2)+P(X=3).
$$

Calculate:

$$
P(X=2)
=
\binom32
\left(\frac13\right)^2
\left(\frac23\right)
=
3\cdot\frac19\cdot\frac23
=
\frac29.
$$

Also,

$$
P(X=3)
=
\binom33
\left(\frac13\right)^3
=
\frac1{27}.
$$

Therefore,

$$
\alpha
=
\frac29+\frac1{27}
=
\frac6{27}+\frac1{27}
=
\frac7{27}.
$$

So,

$$
\boxed{
\alpha=\frac7{27}\approx0.2592592593
}
$$

---

## Type II error and $\beta$

A Type II error means:

> Fail to reject $H_0$ even though $H_1$ is true.

The rejection region is

$$
X\ge2,
$$

so the **non-rejection region** is

$$
X\le1.
$$

Under $H_1$,

$$
p=\frac23.
$$

Therefore,

$$
\beta
=
P_{p=2/3}(X\le1).
$$

Thus,

$$
\beta
=
P(X=0)+P(X=1).
$$

First,

$$
P(X=0)
=
\left(\frac13\right)^3
=
\frac1{27}.
$$

Next,

$$
P(X=1)
=
\binom31
\left(\frac23\right)
\left(\frac13\right)^2.
$$

So,

$$
P(X=1)
=
3\cdot\frac23\cdot\frac19
=
\frac29.
$$

Therefore,

$$
\beta
=
\frac1{27}+\frac29
=
\frac7{27}.
$$

Hence,

$$
\boxed{
\beta=\frac7{27}\approx0.2592592593
}
$$

### Final answer

$$
\boxed{
\alpha=\beta=\frac7{27}\approx0.25926
}
$$

## Why this is a paired t-test

The same golfer supplies both an A distance and a B distance. The two measurements for one golfer are therefore naturally linked.

We do **not** treat all A observations and all B observations as unrelated independent samples.

Instead, we reduce each pair to one difference:

$$

Then we perform a one-sample t-test on the differences.

$$

=
\frac1n\sum_{i=1}^{n}D_i
$$

and

$$
S_D
=
\sqrt{
\frac{
\sum_{i=1}^{n}(D_i-\bar D)^2
}{
n-1
}
}.
$$

The paired t statistic is

$$
\boxed{
T
=
\frac{
\bar D-0

# Question 15 — Monte Carlo Estimation of an Integral

The assignment introduces Monte Carlo integration.

The function is

$$
f(x)
=
\frac{4}{1+x^2},
$$

with

$$
0\le x\le1.

The bounding rectangle is

$$

$$

$$

\approx
\frac{T}{K}(b-a)(M-m)+m(b-a),
$$

where

- $K$ is the number of random points;
- $T$ is the number of those points satisfying

$$
y<f(x).
$$

---

## Why the formula works

The rectangle above the baseline $y=m$ has area

$$
(b-a)(M-m).
$$

The fraction of simulated points that fall below the curve is approximately

$$
\frac{T}{K}.
$$

Therefore the shaded area between the curve and the baseline is approximately

$$
\frac{T}{K}(b-a)(M-m).
$$

But the desired integral also contains the rectangle below the baseline $y=m$, whose area is

$$
m(b-a).
$$

So we add that baseline area.

---

## 15(a) Simulate $K=1,000,000$ points and calculate $T$

40407

\frac4{1+x^2}.
$$

With this fixed seed, the simulation gives

}
$$

Because this is a Monte Carlo simulation, a different random seed will usually give a slightly different $T$.

# Week 07 Mature Theory Cheat Sheet

This section summarizes the concepts needed to solve the complete Week 07 assignment.

---

## 1. Random sample

A random sample

$$
X_1,\ldots,X_n
$$

means the variables are:

1. independent;
2. identically distributed.

This is why Week 07 repeatedly allows us to multiply probabilities and add variances.

---

## 2. Binomial distribution

If

$$
X\sim b(n,p),
$$

then

$$
P(X=k)
=
\binom nk
p^k(1-p)^{n-k}.
$$

Mean:

$$
E[X]=np.
$$

Variance:

$$
\operatorname{Var}(X)=np(1-p).
$$

If

$$
X_1\sim b(n_1,p)
$$

and

$$
X_2\sim b(n_2,p)
$$

are independent, then

$$
\boxed{
X_1+X_2\sim b(n_1+n_2,p).
}
$$

---

## 3. Independence

For independent events/variables,

$$
P(A\cap B)
=
P(A)P(B).
$$

For suitable functions of independent variables,

$$
E[g(X)h(Y)]
=
E[g(X)]E[h(Y)].
$$

This is the key idea behind Questions 1–3.

---

## 4. Second moment

Use

$$
\operatorname{Var}(X)
=
E[X^2]-E[X]^2.
$$

Therefore,

$$
\boxed{
E[X^2]
=
\operatorname{Var}(X)+E[X]^2.
}
$$

This is especially useful for normal random variables.

---

## 5. Mean and variance of sums

Always:

$$
E[X+Y]
=
E[X]+E[Y].
$$

If $X$ and $Y$ are independent:

$$
\operatorname{Var}(X+Y)
=
\operatorname{Var}(X)+\operatorname{Var}(Y).
$$

---

## 6. Sampling distribution of the mean

If each observation has

$$
E[X]=\mu
$$

and

$$
\operatorname{Var}(X)=\sigma^2,
$$

then

$$
E[\bar X]=\mu
$$

and

$$
\operatorname{Var}(\bar X)
=
\frac{\sigma^2}{n}.
$$

The standard error is

$$
\boxed{
SE_{\bar X}
=
\frac{\sigma}{\sqrt n}.
}
$$

If the population itself is normal, $\bar X$ is exactly normal.

---

## 7. Central Limit Theorem

For sufficiently large $n$,

$$
\bar X
\approx
N\left(
\mu,
\frac{\sigma^2}{n}
\right).
$$

For a sum

$$
Y=X_1+\cdots+X_n,
$$

the CLT gives

$$
Y
\approx
N(n\mu,n\sigma^2).
$$

---

## 8. Continuity correction

When approximating a discrete distribution such as a binomial with a normal distribution, move each integer boundary by $0.5$.

Examples:

$$
P(Y=15)
\approx
P(14.5<Y<15.5).
$$

$$
P(Y\le12)
\approx
P(Y<12.5).
$$

$$
P(Y\ge12)
\approx
P(Y>11.5).
$$

This correction is central to Question 8.

---

## 9. Null and alternative hypotheses

The null hypothesis is the default claim:

$$
H_0.
$$

The alternative is what the evidence is trying to support:

$$
H_1.
$$

Examples:

Right-tailed:

$$
H_1:\mu>\mu_0.
$$

Left-tailed:

$$
H_1:p<p_0.
$$

Two-sided:

$$
H_1:\mu\ne\mu_0.
$$

The direction of $H_1$ determines which tail of the sampling distribution is used.

---

## 10. Type I error

A Type I error means:

$$
\boxed{
\text{Reject }H_0
\text{ when }H_0\text{ is true.}
}
$$

Its probability is

$$
\boxed{\alpha}.
$$

The significance level is the allowed Type I error probability.

---

## 11. Type II error

A Type II error means:

$$
\boxed{
\text{Fail to reject }H_0
\text{ when }H_1\text{ is true.}
}
$$

Its probability is

$$
\boxed{\beta}.
$$

When a rejection region is given, find its complement first to identify the Type II region.

---

## 12. p-value

The p-value is:

> The probability, assuming $H_0$ is true, of observing a test statistic at least as extreme in the direction of $H_1$ as the one actually observed.

Decision rule:

$$
p\text{-value}<\alpha
\quad\Longrightarrow\quad
\text{reject }H_0.
$$

Otherwise:

$$
p\text{-value}\ge\alpha
\quad\Longrightarrow\quad
\text{do not reject }H_0.
$$

---

## 13. One-proportion z-test

For

$$
H_0:p=p_0,
$$

use

$$
\boxed{
z=
\frac{
\hat p-p_0
}{
\sqrt{p_0(1-p_0)/n}
}.
}
$$

For a right-tailed alternative,

$$
p\text{-value}
=
1-\Phi(z).
$$

---

## 14. One-sample z-test for a mean

If the population standard deviation $\sigma$ is known,

$$
\boxed{
z
=
\frac{
\bar x-\mu_0
}{
\sigma/\sqrt n
}.
}
$$

Use a z-test, not a t-test.

---

## 15. Paired t-test

Use a paired test when each subject produces two linked measurements.

Define

$$
D_i=X_i-Y_i.
$$

Then test the mean of the differences:

$$
\boxed{
T=
\frac{
\bar D-\mu_{D,0}
}{
S_D/\sqrt n
}.
}
$$

Degrees of freedom:

$$
\boxed{df=n-1}.
$$

Do **not** treat paired observations as two unrelated samples.

---

## 16. Pooled two-sample t-test

For two independent samples under the common-variance model,

$$
S_p^2
=
\frac{
(n-1)S_X^2+(m-1)S_Y^2
}{
n+m-2
}.
$$

Then

$$
\boxed{
T
=
\frac{
(\bar X-\bar Y)-\Delta_0
}{
S_p\sqrt{1/n+1/m}
}.
}
$$

Degrees of freedom:

$$
\boxed{
df=n+m-2.
}
$$

---

## 17. Critical region vs p-value

There are two equivalent decision methods.

### Critical-value method

Reject if the statistic enters the critical region.

Example:

$$
T>t_{0.95,df}.
$$

### p-value method

Reject if

$$
p\text{-value}<\alpha.
$$

A correct analysis should make both methods give the same decision.

---

## 18. Monte Carlo integration

Suppose

$$
m\le f(x)\le M
$$

for

$$
a\le x\le b.
$$

Generate $K$ random points uniformly in the rectangle

$$
[a,b]\times[m,M].
$$

Let $T$ be the number satisfying

$$
y<f(x).
$$

Then

$$
\boxed{
\int_a^bf(x)\,dx
\approx
\frac{T}{K}
(b-a)(M-m)
+
m(b-a).
}
$$

Larger $K$ generally produces a more stable estimate.

# Week 07 Common Mistakes to Avoid

- Forgetting to use independence before multiplying probabilities.
- Forgetting that a random sample implies independent, identically distributed observations.
- Using $\sigma$ instead of $\sigma/\sqrt n$ for a sample mean.
- Forgetting continuity correction in a normal approximation to a binomial.
- Confusing the rejection region with the non-rejection region when calculating $\beta$.
- Using the wrong tail for a p-value.
- Saying “prove $H_0$” when a test only fails to reject it.
- Using an independent-samples test on paired data.
- Reversing the paired difference $A-B$ when the alternative is $\mu_A-\mu_B>0$.
- Forgetting degrees of freedom in a t-test.
- Treating a Monte Carlo estimate as an exact value.
- Expecting the same Monte Carlo count $T$ when a different random seed is used.
